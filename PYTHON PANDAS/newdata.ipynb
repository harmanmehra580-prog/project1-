```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "Name": ["Amit", "Priya", "Rahul", "Sneha", "Arjun",
             "Neha", "Karan", "Meera", "Vikram", "Anjali"],
    "Math": [85, 90, 70, 95, 60, 88, 76, 92, 81, 87],
    "Science": [78, 88, 65, 91, 72, 84, 80, 95, 79, 90],
    "English": [92, 95, 72, 89, 68, 90, 78, 94, 85, 91]
})

df.plot(
    x="Math",
    y="English",
    kind="scatter",
    title="Math vs Science Marks"
)

plt.xlabel("Math Marks")
plt.ylabel("English Marks")
plt.show()
```

```python
import matplotlib.pyplot as plt

x = [-5, -4, -3, -2, -1, 0, 1, 2, 3, 4]
y = [-10, -8, -6, -4, -2, 0, 2, 4, 6, 8]

plt.scatter(x, y)

plt.xlabel("X values")
plt.ylabel("Y values")
plt.title("Negative Values Scatter Chart")

plt.axhline(0)
plt.axvline(0)

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "expenses label"= ['rent','travelling','saving','investment']
}

df = pd.DataFrame(data)

gender_count = df["Gender"].value_counts()

gender_count.plot(
    kind="pie",
    autopct="%1.1f%%",
    title="Student Gender Distribution"
)

plt.ylabel("")
plt.show()
```
